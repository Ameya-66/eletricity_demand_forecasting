```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Keep only the required columns and ensure chronological order
data = df[["Datetime", "PJME_MW"]].copy()
data = data.dropna().sort_values("Datetime").reset_index(drop=True)

# Split chronologically: first 80% for training, last 20% for testing
split_index = int(len(data) * 0.80)

train = data.iloc[:split_index].copy()
test = data.iloc[split_index:].copy()

print("Training records:", len(train))
print("Testing records:", len(test))
print("Training period:", train["Datetime"].min(), "to", train["Datetime"].max())
print("Testing period:", test["Datetime"].min(), "to", test["Datetime"].max())

# Baseline 1: Persistence
# Predict the next hour using the previous hour's actual demand
data["Persistence_Pred"] = data["PJME_MW"].shift(1)

# Baseline 2: Moving average
# Predict using the average demand of the previous 24 hours
data["MA24_Pred"] = (
    data["PJME_MW"].rolling(window=24).mean().shift(1)
)

# Select predictions corresponding to the test period
test_idx = test.index
y_true = data.loc[test_idx, "PJME_MW"].to_numpy()
persistence_pred = data.loc[test_idx, "Persistence_Pred"].to_numpy()
ma24_pred = data.loc[test_idx, "MA24_Pred"].to_numpy()

# Calculate evaluation metrics
def evaluate_model(actual, predicted):
    mask = np.isfinite(actual) & np.isfinite(predicted)
    actual = actual[mask]
    predicted = predicted[mask]

    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mae = mean_absolute_error(actual, predicted)
    mape = np.mean(
        np.abs((actual - predicted) / actual)
    ) * 100

    return rmse, mae, mape

results = []

for name, predictions in [
    ("Persistence", persistence_pred),
    ("Moving Average (24h)", ma24_pred)
]:
    rmse, mae, mape = evaluate_model(y_true, predictions)
    results.append({
        "Model": name,
        "RMSE (MW)": rmse,
        "MAE (MW)": mae,
        "MAPE (%)": mape
    })

baseline_results = pd.DataFrame(results)
display(baseline_results.round(3))
```

```python

# Display the first 7 days of the test period
plot_data = pd.DataFrame({
    "Datetime": data.loc[test_idx, "Datetime"].to_numpy(),
    "Actual Demand": y_true,
    "Persistence": persistence_pred,
    "Moving Average (24h)": ma24_pred
}).dropna()

plot_data = plot_data.head(24 * 7)

plt.figure(figsize=(15, 5))
plt.plot(
    plot_data["Datetime"], plot_data["Actual Demand"],
    label="Actual Demand", linewidth=1.5
)
plt.plot(
    plot_data["Datetime"], plot_data["Persistence"],
    label="Persistence", alpha=0.8
)
plt.plot(
    plot_data["Datetime"], plot_data["Moving Average (24h)"],
    label="Moving Average (24h)", alpha=0.8
)

plt.title("Baseline Models: Actual vs Predicted Electricity Demand")
plt.xlabel("Date")
plt.ylabel("Electricity Demand (MW)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler

# Keep the target column and ensure chronological order
data = df[["Datetime", "PJME_MW"]].copy()
data = data.dropna().sort_values("Datetime").reset_index(drop=True)

# Chronological 80/20 split
split_index = int(len(data) * 0.80)

train_values = data["PJME_MW"].iloc[:split_index].to_numpy().reshape(-1, 1)
test_values = data["PJME_MW"].iloc[split_index:].to_numpy().reshape(-1, 1)

# Fit scaler ONLY on training data to prevent data leakage
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_values)

# Scale test data using the training-set scaler
test_scaled = scaler.transform(test_values)

# Combine scaled values so test predictions can use
# the final 24 observed training hours as context
all_scaled = np.concatenate([train_scaled, test_scaled], axis=0)

LOOKBACK = 24

X_train, y_train = [], []
for i in range(LOOKBACK, split_index):
    X_train.append(all_scaled[i-LOOKBACK:i, 0])
    y_train.append(all_scaled[i, 0])

X_test, y_test = [], []
for i in range(split_index, len(data)):
    X_test.append(all_scaled[i-LOOKBACK:i, 0])
    y_test.append(all_scaled[i, 0])

X_train = np.array(X_train).reshape(-1, LOOKBACK, 1)
y_train = np.array(y_train)

X_test = np.array(X_test).reshape(-1, LOOKBACK, 1)
y_test = np.array(y_test)

print("Training input shape:", X_train.shape)
print("Training target shape:", y_train.shape)
print("Testing input shape:", X_test.shape)
print("Testing target shape:", y_test.shape)
print("Lookback window:", LOOKBACK, "hours")
```

```python

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Reproducibility
tf.keras.utils.set_random_seed(42)

# Build the LSTM model
model = Sequential([
    tf.keras.Input(shape=(LOOKBACK, 1)),
    LSTM(32, return_sequences=True),
    Dropout(0.2),
    LSTM(16),
    Dense(1)
])

# Configure training
model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

# Stop training if validation loss stops improving
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

model.summary()

# Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=256,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stopping],
    verbose=1
)
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Generate LSTM predictions
pred_scaled = model.predict(X_test, verbose=0)

# Convert predictions back to actual electricity demand (MW)
lstm_pred = scaler.inverse_transform(pred_scaled).ravel()
actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
).ravel()

# 2. Calculate baseline predictions on the SAME test period
series = data["PJME_MW"].to_numpy()

# Previous hour's actual demand
persistence_pred = series[split_index - 1:-1]

# Average of the previous 24 hours
ma24_pred = np.array([
    series[i - 24:i].mean()
    for i in range(split_index, len(series))
])

# 3. Define evaluation metrics
def calculate_metrics(actual, predicted):
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mae = mean_absolute_error(actual, predicted)
    mape = np.mean(
        np.abs((actual - predicted) / actual)
    ) * 100

    return rmse, mae, mape

# 4. Compare the models
results = []

for name, pred in [
    ("Persistence", persistence_pred),
    ("Moving Average (24h)", ma24_pred),
    ("LSTM", lstm_pred)
]:
    rmse, mae, mape = calculate_metrics(actual, pred)

    results.append({
        "Model": name,
        "RMSE (MW)": rmse,
        "MAE (MW)": mae,
        "MAPE (%)": mape
    })

comparison = pd.DataFrame(results)
display(comparison.round(3))
```

```python

# Plot the first 7 days of the test period
hours_to_plot = min(24 * 7, len(actual))

dates = data["Datetime"].iloc[
    split_index:split_index + hours_to_plot
]

plt.figure(figsize=(15, 5))

plt.plot(
    dates, actual[:hours_to_plot],
    label="Actual Demand", linewidth=1.5
)
plt.plot(
    dates, persistence_pred[:hours_to_plot],
    label="Persistence", alpha=0.8
)
plt.plot(
    dates, ma24_pred[:hours_to_plot],
    label="Moving Average (24h)", alpha=0.8
)
plt.plot(
    dates, lstm_pred[:hours_to_plot],
    label="LSTM", alpha=0.9
)

plt.title("LSTM vs Baseline Models")
plt.xlabel("Date")
plt.ylabel("Electricity Demand (MW)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, Flatten
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

tf.keras.utils.set_random_seed(42)

# Feed-forward ANN: flatten the previous 24 hours into features
ann_model = Sequential([
    Input(shape=(LOOKBACK, 1)),
    Flatten(),
    Dense(64, activation="relu"),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(1)
])

ann_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

ann_early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

ann_history = ann_model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=256,
    validation_split=0.1,
    shuffle=False,
    callbacks=[ann_early_stopping],
    verbose=1
)

# Predict and convert back to MW
ann_pred_scaled = ann_model.predict(X_test, verbose=0)
ann_pred = scaler.inverse_transform(ann_pred_scaled).ravel()

print("ANN training complete!")
print("Number of ANN predictions:", len(ann_pred))
```

```python

def get_metrics(actual, predicted):
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mae = mean_absolute_error(actual, predicted)
    mape = np.mean(np.abs((actual - predicted) / actual)) * 100
    return rmse, mae, mape

final_results = []

for name, pred in [
    ("Persistence", persistence_pred),
    ("Moving Average (24h)", ma24_pred),
    ("LSTM", lstm_pred),
    ("Feed-forward ANN", ann_pred)
]:
    rmse, mae, mape = get_metrics(actual, pred)
    final_results.append({
        "Model": name,
        "RMSE (MW)": rmse,
        "MAE (MW)": mae,
        "MAPE (%)": mape
    })

final_comparison = pd.DataFrame(final_results)
display(final_comparison.round(3))
```

```python

import matplotlib.pyplot as plt

# Create three comparison charts
metrics_to_plot = [
    ("RMSE (MW)", "Model Comparison: RMSE"),
    ("MAE (MW)", "Model Comparison: MAE"),
    ("MAPE (%)", "Model Comparison: MAPE")
]

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

for ax, (metric, title) in zip(axes, metrics_to_plot):
    ax.bar(final_comparison["Model"], final_comparison[metric])
    ax.set_title(title)
    ax.set_ylabel(metric)
    ax.tick_params(axis="x", rotation=65)
    ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("model_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# Save the results table
final_comparison.to_csv("model_comparison_results.csv", index=False)

print("Saved model_comparison.png")
print("Saved model_comparison_results.csv")
```
